## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python

## 2. The question: decision, action, cost of a wrong call

**Search question:** *Among pages that already rank well enough to be seen (position 1-20, enough impression volume to trust the number), which ones are capturing fewer clicks than other pages sitting at the same position -- and so deserve a human look at their title, meta description, or snippet first?*

**Unit of analysis:** a page (one `content_id`), summarized over its trailing 90-day window -- not a client, not a day, not a query. A client can own many flagged pages; a page can only be reviewed once.

**Output:** a ranked review queue -- one row per flagged page, an opportunity score (how far below its position tier's typical CTR it sits, adjusted so a position-15 page is never compared to a position-2 page), and a reason code the reviewer can read before opening the page.

**Who acts, and what they do:** an SEO/content reviewer with limited weekly review capacity opens the top of the queue and rewrites the title tag, meta description, or on-page snippet structure for the flagged pages -- or checks whether the page's intent actually matches what searchers are looking for.

**Cost of a wrong call:**
- *False positive* (flagged, but CTR was never really recoverable): the reviewer spends time rewriting a title that wasn't the real problem, and a genuinely under-performing page waits longer.
- *False negative* (a real under-performer never surfaces): impressions and rank the site already earned keep going un-clicked, quietly -- lost traffic and likely lost revenue that a five-minute title rewrite might have fixed.
- Because the review itself is cheap, the bigger risk is a *badly ordered* queue -- burning limited attention on low-volume pages while a high-impression page with a real CTR gap sits further down.

**Why data/ML helps at all:** a flat "CTR below X%" rule isn't enough on its own, because CTR is structurally different by position. A useful score has to compare each page only to its own position tier, with a minimum-volume floor. That's more structure than a single if-statement, short of a full model -- which is why this lane calls for a residual/gap score first, with a learned model only if the simple version leaves signal on the table.
```

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


## 4. Careful words: what I can and can't claim

**What this work will be able to say:**
- *Observed*: within this 90-day window, a large share of visible pages sit below their own
  position tier's median CTR -- that's a measured gap, not a guess.
- *Directional / decision-support*: the ranked queue is a prioritization tool for a human
  reviewer with limited time -- "look at this one first" -- not a verdict on any single page.
- The score is relative and tier-adjusted: it says a page under-performs *pages like it* on
  position and volume, nothing stronger.

**What this work will never claim:**
- **Not causal.** A low CTR score is an association with position/volume, not proof that the
  title, meta description, or snippet *caused* the gap -- and rewriting one is not guaranteed to
  close it. Proving that would need an experiment (e.g. an A/B test on the rewritten title), which
  this data can't run.
- **Not a model of Google's ranking algorithm** or of how any search or AI platform decides what
  to show. This data measures clicks and positions that already happened -- not why the algorithm
  chose them.
- **Not a full-warehouse result yet.** These numbers come from the 30,000-row anonymized starter
  slice covering 32 clients over one trailing 90-day window -- useful for judging whether the lane
  is worth pursuing, but not something to generalize to the full ~79M-row warehouse without
  re-checking there.
- **Not free of noise at low volume.** Even with the impressions_90d >= 500 floor, some flagged
  pages will have a low CTR for reasons a title rewrite can't fix (wrong search intent, a
  seasonal dip, a sibling page absorbing the clicks) -- the reason codes exist so a reviewer can
  sanity-check the score before acting on it, not so the score can act alone.
```
