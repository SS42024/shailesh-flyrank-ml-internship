## 5-Minute Demo Outline

**Question (0:00-0:45)**
Content teams can't refresh every published page, so which ones should
they fix first? I asked: can a model flag pages whose search traffic is
declining better than a simple rule?

**Method (0:45-2:00)**
- Data: 30,000 anonymized pages, one row per page, with search and
  engagement metrics over a trailing 90-day window.
- Label: "declining" = trend_direction is "down", a rule-defined proxy,
  not an observed future outcome.
- Model: Random Forest with an 80/20 split.
- Baseline and metric: [Precision@50 of my baseline rule: X], because
  a team can only act on a short list.

**One chart (2:00-3:15)**
Bar chart of precision: all 30 features (1.00), two clean features
(0.65), and the share of pages labeled declining (0.54). The takeaway:
the perfect score vanishes once the label's own ingredients are removed.

**One honest result (3:15-4:15)**
With all 30 numeric features the model scored 1.000, which is too good
to be true. Columns such as trend_pct and the last-30 vs prev-30-day
counts likely define the label, so the model was reading the answer.
With two clean features (content_age_days, impressions_90d) it reached
0.65 precision and 0.612 accuracy, modestly above the 54% base rate.
[Precision@50 of the clean model: X]. Limit: the label is a rule, not
a real future outcome.

**One recommendation (4:15-5:00)**
Use model scores only as decision support to help editors order a
review list. Next step: build the label from a later month's actual
traffic change and test on a held-out period.
