## 5-Minute Demo Outline

**Question (0:00-0:45)**
Content teams can't refresh every published page. Which ones should
they review first? I asked whether a model can rank pages for
refresh better than a simple hand-written rule.

**Method (0:45-2:00)**
- Data: anonymized pages from a search-performance export, one row
  per page, kept if the page had impressions and was at least 90
  days old; [N] rows after cleaning.
- Label: "declining" = trend_direction is "down", a trend flag
  already in the export, not a measured future outcome.
- Models: logistic regression, decision tree, random forest, compared
  with a transparent baseline rule on the same held-out split.
- Metric: Precision@50, because a team can only act on a short list.

**One chart (2:00-3:15)**
Bar chart of Precision@50 for the four methods: baseline 0.24,
logistic regression 0.40, decision tree 0.54, random forest 0.74.
The takeaway: every learned model beats the rule, and the random
forest roughly triples it.

**One honest result (3:15-4:15)**
The random forest's top 50 pages were right 74% of the time versus 24%
for the baseline rule (ROC AUC 0.75, a moderate signal). Limits: the
score rests on only 50 pages, the best model was chosen on the same
test set, and the label is a rule-based proxy, not a measured future
outcome. [Add your Week 2 leakage lesson here only if it applies.]

**One recommendation (4:15-5:00)**
Use the model's ranked list as decision support: editors review the
top pages first and spot-check them. Next step: re-test on a later
time period before relying on it.
