## 5-Minute Demo Outline

**Question (0:00-0:45)**
Content teams can't review every published page. Which ones should
they look at first? I asked whether a model can rank pages that carry
a declining-trend flag better than a simple hand-written rule, as
decision-support for review order. It does not predict whether a
refresh would help.

**Method (0:45-2:00)**
- Data: an anonymized search-performance export, one row per page;
  kept pages with impressions and at least 90 days of age. [N] rows.
- Label: "declining" = trend_direction is "down", a trend flag already
  in the export. It is a proxy, not a measured future outcome.
- Features: 18 numeric and 8 categorical columns. I left out
  trend_pct and the last-30 vs prev-30-day counts because they sit
  too close to the label (using them gave a perfect 1.000 in an early
  test). Other columns may still carry some of the same signal, so I
  ran a leakage check: [drop-one-feature-group result].
- Models: logistic regression, decision tree, random forest, vs. a
  transparent baseline rule, on the same [client-grouped holdout,
  confirmed no client appears in both train and test] split.
- Metric: Precision@50, because a team can only act on a short list.
  Base rate (share of all test pages flagged "down"): [X%].

**One chart (2:00-3:15)**
Bar chart of Precision@50 with a reference line at the base rate:
baseline 0.24, logistic regression 0.40, decision tree 0.54, random
forest 0.74. Observed: the random forest scored about three times the
baseline. The gaps among the three learned models are smaller and
directional, since each score rests on 50 pages (one page moves it by
0.02).

**One honest result (3:15-4:15)**
Measured on the held-out set: 37 of the random forest's top 50 pages
carried the declining flag, versus 12 of 50 for the baseline rule
(ROC AUC 0.75, a moderate signal). Limits:
- The score rests on 50 pages.
- The best model was chosen on the same test set, so 0.74 is likely a
  little optimistic. [Before/after with a separate validation split
  or cross-validation: ___.]
- The label is a proxy.
- Features come from the same 90-day window as the trend label, so the
  model ranks pages that already look weak. It does not forecast the
  future.

This supports prioritizing review order as decision-support. It is
not evidence that refreshing the top-ranked pages will recover traffic.
