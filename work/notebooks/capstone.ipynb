## 5-Minute Demo Outline

**Question (0:00-0:45)**
Content teams can't refresh every published page. Which ones should
they review first? I asked whether a model can rank pages for refresh
better than a simple hand-written rule.

**Method (0:45-2:00)**
- Data: an anonymized search-performance export, one row per page;
  kept pages with impressions and at least 90 days of age. [N] rows.
- Label: "declining" = trend_direction is "down", a trend flag already
  in the export, not a measured future outcome.
- Features: 18 numeric and 8 categorical columns. I deliberately left
  out trend_pct and the last-30 vs prev-30-day counts because they sit
  too close to the label (using them gave a perfect 1.000 in an early
  test).
- Models: logistic regression, decision tree, random forest, vs. a
  transparent baseline rule, on the same [client-holdout] split.
- Metric: Precision@50, because a team can only act on a short list.

**One chart (2:00-3:15)**
Bar chart of Precision@50: baseline 0.24, logistic regression 0.40,
decision tree 0.54, random forest 0.74. Every learned model beats the
rule, and the random forest roughly triples it.

**One honest result (3:15-4:15)**
The random forest's top 50 pages were right 74% of the time versus 24%
for the baseline (ROC AUC 0.75, a moderate signal). Limits: the score
rests on 50 pages; the best model was chosen on the same test set;
the label is a proxy; and the features come from the same 90-day
window as the trend label, so this ranks pages that already look weak
and does not forecast the future.

**One recommendation (4:15-5:00)**
Use the ranked list as decision support: editors review the top pages
first and spot-check them. Next step: re-test on a later time period
before relying on it.
